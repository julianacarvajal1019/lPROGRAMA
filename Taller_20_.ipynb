# **Gestión de Portafolios Financieros**


> Desarrollar un script para ayudar a la empresa a gestionar información sobre las inversiones de los clientes y generar informes sobre el rendimiento de sus portafolios.




# **EJERCICIO 1: Declaración de Variables y Listas**
1. ***Datos de Inversión Inicial***

La empresa ha recopilado los siguientes datos sobre cuatro clientes:

***Clientes:*** Juan Perez, Maria Rodriguez, Carlos Lopez, Ana Gomez

***Montos invertidos***: (en dólares): 10000, 15000, 20000, 18000

***Rendimientos anuales***: (%): 5.5, 6.0, 4.0, 7.5


# **CREAR LISTAS**

```python
Cliente= ["Juan Pérez","Maria Rodríguez","Carlos López","Ana Gómez"]
print(Cliente)
monto_invertido= [10000,15000,20000,18000]
print(monto_invertido)
rendimiento_anual= [5.5,6.0,4.0,7.5]
print(rendimiento_anual)
```

# **CREAR TUPLAS DE INFORMACIÓN**

# **EJERCICIO 2: Uso de Diccionarios y Condicionales**
1. ***Almacenar los datos en un diccionario***



```python
portafolio = {
    "Juan Pérez": {"monto_invertido": 10000, "rendimiento_anual": 5.5},
    "Maria Rodríguez": {"monto_invertido": 15000, "rendimiento_anual": 6.0},
    "Carlos López": {"monto_invertido": 20000, "rendimiento_anual": 4.0},
    "Ana Gómez": {"monto_invertido": 18000, "rendimiento_anual": 7.5}
}
```

2. ***Evaluación del Rendimiento de los Clientes***

Qué clientes han tenido un rendimiento superior al 5%.

```python
for Cliente, datos in portafolio.items():
    if datos["rendimiento_anual"] > 5:
        print(f"{Cliente}' tiene un rendimiento anual de {datos['rendimiento_anual']}%")

    elif datos["rendimiento_anual"] < 5:
         print(f"{Cliente} tiene un rendimiento anual menor al 5%")
```

# **EJERCICIO 3: Cálculos con Ciclos**
1. ***Calcular el Rendimiento Total del Portafolio***

La empresa desea calcular el rendimiento total del portafolio, que se define como la suma de todos los rendimientos en dólares
  


```python
rendimiento_total = 0
for Cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento_anual"]) / 100
    rendimiento_total += rendimiento_cliente

print(f"El rendimiento total del portafolio es: ${rendimiento_total:.2f}")
```

2. ***Identificar al Cliente con Mayor Rendimiento***

La empresa necesita saber qué cliente ha obtenido el mayor rendimiento en términos absolutos.

```python
Max_rendimiento = 0
Cliente_Max_rendimiento = ""

for Cliente, datos in portafolio.items():
    rendimiento_cliente = (datos["monto_invertido"] * datos["rendimiento_anual"]) / 100

    if rendimiento_cliente > Max_rendimiento:
        Max_rendimiento = rendimiento_cliente
        Cliente_Max_rendimiento = Cliente

print(f"El Cliente con el mayor rendimiento es {Cliente_Max_rendimiento} con un rendimiento de ${Max_rendimiento:.2f}")
```
