# **TIR**

```python
import numpy_financial as npf

# Definimos la función para calcular la Tasa Interna de Retorno (TIR) de un proyecto
def calcular_tir(row):
    flujos = [row['Inversion_Inicial']] + [row.get(f'Año_{t}', 0) for t in range(1, 11)]  # Años 1 a 10
    return npf.irr(flujos)

# Aplicamos la función 'calcular_tir' a cada fila
df['TIR'] = df.apply(calcular_tir, axis=1)

# Mostramos los resultados, que consisten en el nombre del proyecto y su respectiva TIR
print("Tasa interna de retorno por proyecto:")
print(df[['Proyecto', 'TIR']])
```

# **Análisis de Proyectos según el Valor Presente Neto (VPN) y la Tasa Interna de Retorno (TIR)**
**Distribución del VPN:**
La distribución del Valor Presente Neto muestra una concentración de proyectos con valores bajos o negativos, y solo una parte significativa con VPN positivos altos. Esto indica que muchos proyectos no están generando el retorno esperado sobre la inversión inicial, mientras que unos pocos tienen un desempeño sobresaliente.

**Distribución de la TIR:**
La TIR, al igual que el VPN, muestra una dispersión importante. Hay varios proyectos con tasas bajas o incluso negativas, y unos pocos con TIR elevadas. Esto refleja una variedad en la rentabilidad relativa, y confirma que no todos los proyectos están siendo eficientes en generar retorno proporcional al capital invertido.
